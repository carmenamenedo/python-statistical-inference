Ahora vamos a utilizar la base de datos Iris, muy popularizada por un artículo de Fisher [1].
Esta base de datos fue utilizada para demostrar que estas medidas en ciertas flores del iris podrían utilizarse para diferenciar entre especies de plantas iris. Contiene 3 clases de 50 casos cada una, donde cada clase se refiere a un tipo de planta iris [2].

Los atributos  del fichero iris.data.mio.cvs son:
    #Longitud del sépalo en cm (Sepal.Length)
    #Ancho del sépalo en cm (Sepal.Width)
    #Longitud del pétalo en cm (Petal.Length)
    #Ancho del pétalo en cm (Pepal.Width)

Clase (Species):
    #Iris Setosa (0)
    #Iris Versicolour (1)
    #Iris Virginica (2)
    

Bibliografía:
    #[1] Fisher, R. A. (1936). The use of multiple measurements in taxonomic problems. Annals of eugenics, 7(2), 179-188.
    #[2] Lichman, M. (2007). UCI machine learning repository (2013). URL archive.ics.uci.edu/ml. Irvine, CA: University of California, School of Information and Computer Science.

```python
# vamos a hacer un análisis exploratorio muy básico de las diferentes características de cada patrón
# Vamos a sacar la cabecera del fichero y los 10 primeros datos
file_name = r'iris.data.mio.csv' 
iris = pd.read_csv(file_name, sep=',')
iris.head(10)

# Calculo del número de muestras
n_muestras = len(iris)
print("Número de muestras: {}".format(n_muestras))
```

```python
# Alguna estadísticas de Iris
estadistica = lambda a : pd.Series([a.count(),np.max(a), np.min(a), (np.max(a)-np.min(a)), np.mean(a), \
                              np.percentile(a, 25), np.median(a), np.percentile(a, 75),(float((stats.mode(a))[0])),\
                              np.std(a), (np.std(a) / abs(np.mean(a))), np.std(a)**2, a.skew(), a.kurt()], \
                             index = ['Numero de mediciones','Medición máxima', 'Medición mínima', 'Recorrido', \
                                      'Media aritmética', '1er Cuartil','Mediana', '3er Cuartil','Moda',\
                                     'Desviación estándar', 'Coeficiente de variación', 'Varianza',\
                                     'Coeficiente de asimetría','Curtosis'])
```

```python
# Clase 0
print("CLASE-0")
iris.groupby('class').get_group(0).drop(['class'], axis = 1).apply(estadistica).round(2)
```

```python
# Clase 1
print("CLASE-1")
iris.groupby('class').get_group(1).drop(['class'], axis = 1).apply(estadistica).round(2)
```

```python
# Clase 2
print("CLASE-2")
iris.groupby('class').get_group(2).drop(['class'], axis = 1).apply(estadistica).round(2)
```

```python
# Saco las clases
setosa_mask = iris['class'] == 0
versicolour_mask = iris['class'] == 1
virginica_mask = iris['class'] == 2

# Separo el data frame para cada especie de iris

iris_setosa = iris[setosa_mask]
iris_versicolour = iris[versicolour_mask]
iris_virginica = iris[virginica_mask]

# Representación longitud del sepalo

xfSize=15
yfSize=10
plt.figure(figsize=(xfSize, yfSize))

plt.subplot(2, 2, 1)
plt.title('%s' % 'Longitud del sépalo')
plt.xlabel('Longitud')
plt.ylabel('Frecuencia absoluta')

plt.hist(iris_setosa['sepal_length'], 10, alpha = 0.5, label='Setosa')
plt.hist(iris_versicolour['sepal_length'], 10, alpha = 0.5, label='Versicolour')
plt.hist(iris_virginica['sepal_length'], 10, alpha = 0.5, label='Virginica')

plt.legend(loc = 'upper right')

# Representacion anchura del sepalo

plt.subplot(2, 2, 2)

plt.title('%s' % 'Anchura del sépalo')
plt.xlabel('Anchura')
plt.ylabel('Frecuencia absoluta')

plt.hist(iris_setosa['sepal_width'], 10, alpha = 0.5, label='Setosa')
plt.hist(iris_versicolour['sepal_width'], 10, alpha = 0.5, label='Versicolour')
plt.hist(iris_virginica['sepal_width'], 10, alpha = 0.5, label='Virginica')

plt.legend(loc = 'upper right')


# Representacion longitud del petalo

plt.subplot(2, 2, 3)
plt.title('%s' % 'Longitud del pétalo')
plt.xlabel('Longitud')
plt.ylabel('Frecuencia absoluta')

plt.hist(iris_setosa['petal_length'], 10, alpha = 0.5, label='Setosa')
plt.hist(iris_versicolour['petal_length'], 10, alpha = 0.5, label='Versicolour')
plt.hist(iris_virginica['petal_length'], 10, alpha = 0.5, label='Virginica')

plt.legend(loc = 'upper right')


# Representacion anchura del petalo

plt.subplot(2, 2, 4)
plt.title('%s' % 'Anchura del pétalo')
plt.xlabel('Anchura')
plt.ylabel('Frecuencia absoluta')

plt.hist(iris_setosa['petal_width'], 5, alpha = 0.5, label='Setosa')
plt.hist(iris_versicolour['petal_width'], 5, alpha = 0.5, label='Versicolour')
plt.hist(iris_virginica['petal_width'], 5, alpha = 0.5, label='Virginica')

plt.legend(loc = 'upper right')

plt.show()
```

```python
# basic plot
# Con esta función de python los “bigotes” son las líneas que van desde cada extremo 
# del rectángulo central hasta el valor más alejado no atípico, 
# es decir, que está dentro del intervalo (LI, LS) comentado de las transparencias del curso.

_=plt.boxplot(dataM[:,1])
```

```python
# basic plot
# cambiar el el tamño 1.5 para considerar dato atípico
# Q3 + whis*IQR, IQR = interquartile range, Q3-Q1
# Q1 - whis*IQR, IQR = interquartile range, Q3-Q1
_=plt.boxplot(dataM[:,1],whis=0.5)
```

```python
# Pintar varios Boxplots
xfSize=15
yfSize=5
plt.figure(figsize=(xfSize, yfSize))

plt.subplot(1, 3, 1);
plt.title('%s' % 'Girth')
_ =plt.boxplot(dataM[:,1])
plt.subplot(1, 3, 2);
plt.title('%s' % 'Height')
_ =plt.boxplot(dataM[:,2])
plt.subplot(1, 3, 3);
plt.title('%s' % 'Volume')
_ =plt.boxplot(dataM[:,3])
```

```python
# ============================================================
# DIAGRAMA DE CAJA CON MUESCA (NOTCHED BOXPLOT)
# ============================================================

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

# Representamos el grosor de los árboles (Girth)
# notch=True añade una muesca alrededor de la mediana
plt.boxplot(dataM[:, 1], notch=True)

plt.title("Diagrama de caja con muesca - Girth")
plt.ylabel("Girth")
plt.grid(axis="y")

plt.show()
```

```python
# ============================================================
# DIAGRAMA DE CAJA HORIZONTAL - GIRTH
# ============================================================

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

# Representamos el grosor de los árboles
# notch=False: sin muesca
# vert=False: orientación horizontal
# sym='rs': valores atípicos representados con cuadrados rojos

plt.boxplot(
    dataM[:, 1],
    notch=False,
    sym='rs',
    vert=False
)

plt.title("Diagrama de caja horizontal - Girth")
plt.xlabel("Girth")
plt.grid(axis="x")

plt.show()
```

```python
# Box plots, visualizar mejor los atípicos
fSize=8
plt.figure(figsize=(fSize, fSize))

# Box plots de las cuatro columnas
# ojo con el operador colon,: , ya que selecciona de esta forma [inicio,fin), intervalo cerado abierto.
bp = plt.boxplot(dataM[:,1:4])

# visualizar mejor los atípicos
for flier in bp['fliers']:
    flier.set(marker='o', color='red', alpha=0.5)
```

# 7) El cuarteto de Anscombe y la regresión lineal.

[El cuarteto de Anscombe](https://en.wikipedia.org/wiki/Anscombe's_quartet) es una agrupación de cuatro conjuntos de datos que tienen las mismas propiedades estadísticas, siendo los datos completamente diferentes [1]. Estos datos los tenéis en el fichero CuartetoAnscombe.csv. Cada uno de los conjuntos en el fichero consiste de once puntos (x, y) construidos por el estadístico F. J. Anscombe [1]. Las columnas 0 y 1 corresponden al primer conjunto, las 2 y 3 al segundo conjunto, y así sucesivamente. Este conjunto de ficheros una demostración de la importancia de mirar gráficamente un conjunto de datos antes de analizarlos.

    #[1] F.J. Anscombe, "Graphs in Statistical Analysis," American Statistician, 27 (febrero de 1973), 17-21.

```python
# Regresión lineal con polyfit
# X e Y serán el primer conjunto del cuarteto de Anscombe 

X = dataM[:,0]
Y = dataM[:,1]

# Calculamos los coeficientes del ajuste (a X + b)
b, a = np.polyfit(X, Y, 1)
# Calculamos el coeficiente de correlación
r = np.corrcoef(X, Y)

# Dibujamos los datos para poder visualizarlos y ver si sería lógico 
# considerar el ajuste usando un modelo lineal
plt.plot(X, Y, 'o')
plt.xlim(np.min(X) -1, np.max(X) +1)
plt.ylim(np.min(Y) -1, np.max(Y) +1)
plt.plot(X, a + b * X)
plt.text(4, 10, 'r = {0:2.3f}'.format(r[0,1]))
_ =plt.text(4, 9, 'Y = {0:2.3f} X + {1:2.3f}'.format(b, a))
```

```python
# Regresión lineal paso a paso: y =a+bx, comparada con polyfit
# X e Y serán el primer conjunto del cuarteto de Anscombe 

X = dataM[:,0]
Y = dataM[:,1]

# Primero calculamos la media de X e Y
xmean = X.mean()
ymean = Y.mean()

# Como A depende de B primero hemos de obtener B
B = np.sum((X - xmean) * (Y - ymean)) / np.sum((X - xmean)**2)
A = ymean - B * xmean

# Calculamos los valores usando numpy.polyfit
b, a = np.polyfit(X, Y, 1)

# Mostramos resultados
print('ajuste paso a paso, a, b = ', A, B)
print('ajuste con polyfit, a, b = ', a, b)
```

```python
# ============================================================
# TURNO 3
# Convergencia de la probabilidad de obtener un número par
# al lanzar un dado
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# Fijamos una semilla para poder reproducir los resultados
np.random.seed(1111111)

# Número de lanzamientos
n = 10000

# Simulamos n lanzamientos de un dado:
# los posibles resultados son 1, 2, 3, 4, 5 y 6
lanzamientos = np.random.randint(1, 7, size=n)

# Comprobamos si cada lanzamiento es par
# True = ha salido par
# False = ha salido impar
es_par = (lanzamientos % 2 == 0)

# Calculamos el número acumulado de resultados pares
pares_acumulados = np.cumsum(es_par)

# Número de lanzamientos realizados: 1, 2, ..., n
numero_lanzamientos = np.arange(1, n + 1)

# Frecuencia relativa acumulada de resultados pares
probabilidad_estimada = pares_acumulados / numero_lanzamientos

# Mostramos algunos resultados
print("Número de lanzamientos =", n)
print("Número de resultados pares =", pares_acumulados[-1])
print("Probabilidad experimental de obtener par =", probabilidad_estimada[-1])
print("Probabilidad teórica de obtener par =", 0.5)

# Representación gráfica
plt.figure(figsize=(10, 5))

plt.plot(numero_lanzamientos, probabilidad_estimada,
         label="Probabilidad experimental")

plt.axhline(y=0.5, linestyle="--",
            label="Probabilidad teórica = 0.5")

plt.xlabel("Número de lanzamientos")
plt.ylabel("Frecuencia relativa de obtener par")
plt.title("Convergencia de la probabilidad de obtener un número par")
plt.legend()
plt.grid()

plt.show()
```

# Tu turno 4: Probar el TCL para una distribución de Poisson

```python
# ============================================================
# TURNO 4
# Teorema Central del Límite para una distribución de Poisson
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parámetro de la distribución de Poisson
Lambda = 4

# Número de muestras que vamos a generar
numero_muestras = 10000

# Diferentes tamaños de muestra para observar la convergencia
tamanos = [1, 5, 30, 100]

plt.figure(figsize=(12, 8))

for i, n in enumerate(tamanos):

    # Generamos numero_muestras muestras de tamaño n
    muestras = np.random.poisson(
        lam=Lambda,
        size=(numero_muestras, n)
    )

    # Calculamos la media de cada muestra
    medias_muestrales = np.mean(muestras, axis=1)

    # Parámetros teóricos de la distribución de la media
    media_teorica = Lambda
    desviacion_teorica = np.sqrt(Lambda / n)

    # Dibujamos el histograma
    plt.subplot(2, 2, i + 1)

    plt.hist(
        medias_muestrales,
        bins=30,
        density=True,
        alpha=0.7
    )

    # Normal teórica predicha por el TCL
    x = np.linspace(
        medias_muestrales.min(),
        medias_muestrales.max(),
        300
    )

    y = norm.pdf(
        x,
        loc=media_teorica,
        scale=desviacion_teorica
    )

    plt.plot(x, y, linewidth=2)

    plt.title("Tamaño de muestra n = " + str(n))
    plt.xlabel("Media muestral")
    plt.ylabel("Densidad")
    plt.grid()

plt.tight_layout()
plt.show()
```

### Interpretación del Turno 4: Teorema Central del Límite para una distribución de Poisson

Se ha utilizado una variable aleatoria con distribución de Poisson de parámetro λ = 4. Para comprobar el Teorema Central del Límite se han generado 10000 muestras para diferentes tamaños muestrales: n = 1, 5, 30 y 100. Para cada muestra se ha calculado su media.

En una distribución de Poisson se cumple que:

E(X) = λ = 4

Var(X) = λ = 4

Por tanto, para la distribución muestral de la media:

E(X̄) = λ = 4

Var(X̄) = λ/n = 4/n

Al observar las gráficas se comprueba que para n = 1 la distribución conserva la forma discreta característica de una Poisson y todavía no presenta una forma normal.

Cuando aumenta el tamaño de la muestra a n = 5, la distribución de las medias comienza a adquirir una forma aproximadamente acampanada.

Para n = 30 la distribución de las medias muestrales presenta ya una clara aproximación a una distribución normal, y para n = 100 el histograma prácticamente coincide con la curva normal teórica representada sobre él.

Además, todas las distribuciones permanecen centradas aproximadamente en 4, que es la media teórica de la distribución de Poisson original. Sin embargo, al aumentar n disminuye la dispersión de las medias muestrales, ya que su varianza es λ/n.

Por tanto, los resultados obtenidos permiten comprobar gráficamente el Teorema Central del Límite: aunque la población original sigue una distribución de Poisson, la distribución de las medias muestrales tiende a una distribución normal conforme aumenta el tamaño de la muestra.

```python
# ============================================================
# TURNO 6A
# Distribución de Weibull:
# tasa de fallo decreciente, constante y creciente
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# Valores de tiempo positivos
t = np.linspace(0.01, 5, 1000)

# Parámetro de escala
h = 1

# Elegimos tres valores del parámetro c
# c < 1  -> tasa de fallo decreciente
# c = 1  -> tasa de fallo constante
# c > 1  -> tasa de fallo creciente

c_decreciente = 0.5
c_constante = 1
c_creciente = 2

# Función utilizada en el código de Weibull de la práctica
def weibull(t, h, c):
    return (h * (t**(c-1))) * np.exp(-(h/c)*(t**c))

# Calculamos las tres distribuciones
W_decreciente = weibull(t, h, c_decreciente)
W_constante = weibull(t, h, c_constante)
W_creciente = weibull(t, h, c_creciente)

# Representamos los resultados
plt.figure(figsize=(15, 4))

# ------------------------------------------------------------
# 1. Tasa de fallo decreciente
# ------------------------------------------------------------
plt.subplot(1, 3, 1)
plt.plot(t, W_decreciente)
plt.title("Weibull: tasa de fallo decreciente\nc = 0.5")
plt.xlabel("Tiempo t")
plt.ylabel("f(t)")
plt.grid()

# ------------------------------------------------------------
# 2. Tasa de fallo constante
# ------------------------------------------------------------
plt.subplot(1, 3, 2)
plt.plot(t, W_constante)
plt.title("Weibull: tasa de fallo constante\nc = 1")
plt.xlabel("Tiempo t")
plt.ylabel("f(t)")
plt.grid()

# ------------------------------------------------------------
# 3. Tasa de fallo creciente
# ------------------------------------------------------------
plt.subplot(1, 3, 3)
plt.plot(t, W_creciente)
plt.title("Weibull: tasa de fallo creciente\nc = 2")
plt.xlabel("Tiempo t")
plt.ylabel("f(t)")
plt.grid()

plt.tight_layout()
plt.show()
```

### Interpretación de la distribución de Weibull

Se ha estudiado el comportamiento de la distribución de Weibull modificando el parámetro de forma c.

- Para c = 0.5 < 1, la tasa de fallo es decreciente. Esto representa situaciones en las que la probabilidad instantánea de fallo es mayor al principio y disminuye con el tiempo.
- Para c = 1, la tasa de fallo es constante. En este caso particular, la distribución de Weibull se reduce a una distribución exponencial.
- Para c = 2 > 1, la tasa de fallo es creciente. Este comportamiento puede representar procesos de envejecimiento o desgaste, en los que la posibilidad de fallo aumenta con el tiempo.

Por tanto, el parámetro c determina el comportamiento de la tasa de fallo de la distribución de Weibull.

```python
# ============================================================
# TURNO 6B
# Distribución t de Student
# Convergencia hacia una Normal N(0,1)
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, norm

# Para poder reproducir exactamente los resultados
np.random.seed(1111111)

# Número de observaciones simuladas
muestras = 100000

# Diferentes grados de libertad
grados_libertad = [1, 2, 5, 10, 30, 100]

# Valores de x para representar las densidades teóricas
x = np.linspace(-5, 5, 1000)

plt.figure(figsize=(15, 10))

for i, n in enumerate(grados_libertad):

    # Simulamos una t de Student
    datos_t = np.random.standard_t(n, size=muestras)

    # Media y varianza experimentales
    media_exp = np.mean(datos_t)
    var_exp = np.var(datos_t)

    # Varianza teórica:
    # para n > 2 es n/(n-2)
    if n > 2:
        var_teorica = n / (n - 2)
    else:
        var_teorica = np.inf

    print("======================================")
    print("Grados de libertad =", n)
    print("Media experimental =", media_exp)
    print("Varianza experimental =", var_exp)
    print("Media teórica =", 0 if n > 1 else "no definida")
    print("Varianza teórica =", var_teorica)

    # Representación
    plt.subplot(2, 3, i+1)

    # t de Student teórica
    plt.plot(x, t.pdf(x, df=n), label="t-Student")

    # Normal estándar
    plt.plot(x, norm.pdf(x, 0, 1), '--', label="Normal N(0,1)")

    plt.title("Grados de libertad = " + str(n))
    plt.xlabel("x")
    plt.ylabel("Densidad")
    plt.grid()
    plt.legend()

plt.tight_layout()
plt.show()
```

### Interpretación del Turno 6

#### A) Distribución de Weibull y distribución geométrica

La distribución de Weibull permite representar diferentes comportamientos de la tasa de fallo dependiendo de su parámetro de forma.

- Para un parámetro de forma menor que 1, la tasa de fallo es decreciente.
- Para un parámetro de forma igual a 1, la tasa de fallo es constante.
- Para un parámetro de forma mayor que 1, la tasa de fallo es creciente.

El caso de tasa de fallo constante es especialmente importante, ya que representa la ausencia de envejecimiento: la probabilidad de fallo no depende del tiempo transcurrido.

En el caso discreto, esta propiedad conduce a la distribución geométrica. Se ha representado la distribución geométrica con una probabilidad constante de fallo $p=0.3$, cuya función de probabilidad es:

$$
P(X=k)=p(1-p)^{k-1}
$$

Por tanto, la distribución geométrica representa el número de periodos que transcurren hasta que se produce el primer fallo cuando la probabilidad de fallo permanece constante.

---

#### B) Distribución t de Student

La variable t de Student puede construirse a partir de una variable normal estándar $Z$ y una variable chi-cuadrado independiente con $n$ grados de libertad:

$$
T=\frac{Z}{\sqrt{\chi_n^2/n}}
$$

La distribución t de Student es simétrica alrededor de cero, pero para pocos grados de libertad presenta colas más pesadas que la distribución normal.

Al aumentar los grados de libertad, la variable chi-cuadrado dividida entre sus grados de libertad se concentra alrededor de 1:

$$
\frac{\chi_n^2}{n}\rightarrow 1
$$

y, por tanto:

$$
T\rightarrow Z
$$

Es decir, la distribución t de Student converge hacia una distribución normal estándar $N(0,1)$.

Esto se observa en las gráficas obtenidas: para pocos grados de libertad existen diferencias claras entre la t de Student y la normal, mientras que para $n=30$ y especialmente para $n=100$ ambas curvas prácticamente se superponen.

La media teórica de una t de Student es:

$$
E(T)=0,\qquad n>1
$$

y su varianza es:

$$
Var(T)=\frac{n}{n-2},\qquad n>2
$$

Para $n=1$, la media no está definida y para $n\leq2$ la varianza no es finita.

Los resultados de la simulación confirman los valores teóricos. Por ejemplo:

- Para $n=5$, se obtiene una varianza experimental aproximadamente igual a 1.645, frente al valor teórico 1.667.
- Para $n=10$, la varianza experimental es aproximadamente 1.245, frente al valor teórico 1.25.
- Para $n=30$, la varianza experimental es aproximadamente 1.079, frente al valor teórico 1.071.
- Para $n=100$, la varianza experimental es aproximadamente 1.013, frente al valor teórico 1.020.

Además, las medias experimentales para grados de libertad suficientemente grandes son muy próximas a cero.

Finalmente:

$$
\lim_{n\to\infty}E(T)=0
$$

y

$$
\lim_{n\to\infty}Var(T)
=
\lim_{n\to\infty}\frac{n}{n-2}
=1
$$

Por tanto, tanto gráficamente como mediante la media y la varianza se comprueba que, cuando los grados de libertad tienden a infinito, la distribución t de Student converge hacia una normal estándar $N(0,1)$.

## Demostrar gráficamente como la identificación del modelo de Poisson se realiza de una manera más correcta cuando se cambian los parámetros de la muestra. Ayúdate para demostrar esto del ajuste de una regresión lineal, para obtener la pendiente, ordenada el origen y así la calidad del ajuste.

```python
# ============================================================
# TURNO 7
# Identificación de una distribución de Poisson
# Efecto del tamaño de la muestra
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.special import factorial

# Para poder reproducir los mismos resultados
np.random.seed(1111111)

# Parámetro verdadero de la distribución Poisson
Lambda = 1

# Diferentes tamaños de muestra que vamos a estudiar
tamanos = [30, 100, 1000, 10000]

# Creamos una figura con 4 gráficos
plt.figure(figsize=(12, 9))

for i, muestras in enumerate(tamanos):

    # --------------------------------------------------------
    # 1. Generamos una muestra procedente de Poisson(lambda=1)
    # --------------------------------------------------------
    x_p = np.random.poisson(Lambda, size=muestras)

    # --------------------------------------------------------
    # 2. Contamos cuántas veces aparece cada valor
    # --------------------------------------------------------
    valores, frecuencias = np.unique(x_p, return_counts=True)

    # --------------------------------------------------------
    # 3. Transformación utilizada para identificar una Poisson
    #
    # Si X sigue una Poisson:
    #
    # P(X=x) = exp(-lambda) * lambda^x / x!
    #
    # Tomando logaritmos:
    #
    # log(P(X=x) * x!) = -lambda + x*log(lambda)
    #
    # que tiene forma de recta:
    #
    # Y = a + bX
    # --------------------------------------------------------

    probabilidades = frecuencias / muestras

    X = valores
    Y = np.log(probabilidades) + np.log(factorial(X))

    # --------------------------------------------------------
    # 4. Ajustamos una regresión lineal
    # --------------------------------------------------------
    pendiente, ordenada = np.polyfit(X, Y, 1)

    # Correlación
    r = np.corrcoef(X, Y)[0, 1]

    # Coeficiente de determinación
    R2 = r**2

    # --------------------------------------------------------
    # 5. Mostramos los resultados numéricos
    # --------------------------------------------------------
    print("========================================")
    print("Tamaño de muestra =", muestras)
    print("Lambda verdadero =", Lambda)
    print("Pendiente =", pendiente)
    print("Ordenada en el origen =", ordenada)
    print("Correlación r =", r)
    print("Coeficiente de determinación R² =", R2)

    # --------------------------------------------------------
    # 6. Representación gráfica
    # --------------------------------------------------------
    plt.subplot(2, 2, i+1)

    # Datos transformados
    plt.scatter(X, Y, label="Datos")

    # Recta de regresión
    plt.plot(X, ordenada + pendiente*X,
             label="Regresión lineal")

    plt.title(
        "Poisson λ = {}\nn = {}, R² = {:.4f}".format(
            Lambda, muestras, R2
        )
    )

    plt.xlabel("x")
    plt.ylabel("log(P(X=x)) + log(x!)")
    plt.grid()
    plt.legend()

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# TURNO 7 - IDENTIFICACIÓN DE UNA DISTRIBUCIÓN DE POISSON
# Estudio para lambda = 5 con distintos tamaños de muestra
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gammaln

# Semilla para poder reproducir los resultados
np.random.seed(1111111)

# Parámetro de la distribución de Poisson
Lambda = 5

# Distintos tamaños de muestra
tamanos = [100, 300, 1000, 10000]

# Creamos la figura con 4 gráficos
plt.figure(figsize=(12, 9))


# ============================================================
# REPETIMOS EL ESTUDIO PARA CADA TAMAÑO DE MUESTRA
# ============================================================

for i, n in enumerate(tamanos):

    # --------------------------------------------------------
    # 1. Generamos datos de una distribución Poisson
    # --------------------------------------------------------
    x_p = np.random.poisson(Lambda, size=n)


    # --------------------------------------------------------
    # 2. Calculamos los valores observados y sus frecuencias
    # --------------------------------------------------------
    valores, frecuencias = np.unique(x_p, return_counts=True)

    # Pasamos las frecuencias a probabilidades experimentales
    probabilidades = frecuencias / n


    # --------------------------------------------------------
    # 3. Transformación para identificar una Poisson
    #
    # Para una Poisson:
    #
    # P(X=x) = exp(-lambda) * lambda^x / x!
    #
    # Tomando logaritmos:
    #
    # log(P(X=x)) = -lambda + x*log(lambda) - log(x!)
    #
    # Por tanto:
    #
    # log(P(X=x)) + log(x!)
    #       = -lambda + x*log(lambda)
    #
    # Esto tiene la forma:
    #
    # Y = a + bX
    #
    # donde:
    # pendiente b = log(lambda)
    # ordenada a = -lambda
    # --------------------------------------------------------

    # Calculamos log(x!)
    # gammaln(x+1) = log(x!)
    log_factorial = gammaln(valores + 1)

    # Variable transformada Y
    Y = np.log(probabilidades) + log_factorial

    # Variable X
    X = valores


    # --------------------------------------------------------
    # 4. REGRESIÓN LINEAL
    # --------------------------------------------------------

    # Calculamos pendiente y ordenada en el origen
    pendiente, ordenada = np.polyfit(X, Y, 1)

    # Valores estimados mediante la recta
    Y_pred = pendiente * X + ordenada

    # Coeficiente de correlación
    r = np.corrcoef(X, Y)[0, 1]

    # Coeficiente de determinación
    R2 = r**2


    # --------------------------------------------------------
    # 5. MOSTRAMOS LOS RESULTADOS
    # --------------------------------------------------------

    print("============================================")
    print("Poisson lambda =", Lambda)
    print("Tamaño de muestra n =", n)

    print("Pendiente experimental =", pendiente)
    print("Pendiente teórica log(lambda) =", np.log(Lambda))

    print("Ordenada experimental =", ordenada)
    print("Ordenada teórica -lambda =", -Lambda)

    print("Correlación r =", r)
    print("Coeficiente de determinación R² =", R2)


    # --------------------------------------------------------
    # 6. REPRESENTACIÓN GRÁFICA
    # --------------------------------------------------------

    plt.subplot(2, 2, i + 1)

    # Datos experimentales
    plt.scatter(X, Y, label="Datos")

    # Recta de regresión
    plt.plot(X, Y_pred, label="Regresión lineal")

    plt.xlabel("x")
    plt.ylabel("log(P(X=x)) + log(x!)")

    plt.title(
        "Poisson λ = {}\n"
        "n = {}, R² = {:.4f}".format(Lambda, n, R2)
    )

    plt.grid()
    plt.legend()


# Ajustamos los gráficos
plt.tight_layout()

# Mostramos los cuatro gráficos
plt.show()


# ============================================================
# VALORES TEÓRICOS ESPERADOS
# ============================================================

print("\n============================================")
print("VALORES TEÓRICOS PARA lambda =", Lambda)
print("Pendiente teórica = log(lambda) =", np.log(Lambda))
print("Ordenada teórica =", -Lambda)
print("============================================")
```

Partiendo de la función de probabilidad de una variable aleatoria Poisson:

$$

$$

y tomando logaritmos se obtiene:

Para $\lambda=5$, los valores teóricos son:

a=-5
$$

Los resultados experimentales obtenidos son próximos a estos valores teóricos. Además, los coeficientes de determinación $R^2$ obtenidos son muy próximos a 1, lo que indica que la transformación presenta un ajuste lineal muy elevado.

```python
# ============================================================
# TURNO 8
# DISTRIBUCIÓN MUESTRAL DE LA MEDIA DE UNA BINOMIAL
# Estudio de la convergencia de la media y la varianza
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Semilla para poder reproducir los resultados
np.random.seed(1111111)

# ------------------------------------------------------------
# 1. PARÁMETROS DE LA DISTRIBUCIÓN BINOMIAL
# ------------------------------------------------------------

# X ~ Binomial(N, p)
N = 20
p = 0.2

# Número de muestras que vamos a generar para cada caso
numero_muestras = 10000

# Distintos tamaños de cada muestra
tamanos_muestra = [1, 5, 30, 100]


# ------------------------------------------------------------
# 2. PARÁMETROS TEÓRICOS DE LA VARIABLE BINOMIAL
# ------------------------------------------------------------

# Media de una Binomial
mu = N * p

# Varianza de una Binomial
varianza = N * p * (1 - p)

print("==============================================")
print("DISTRIBUCIÓN ORIGINAL: BINOMIAL")
print("N =", N)
print("p =", p)
print("Media teórica de X =", mu)
print("Varianza teórica de X =", varianza)
print("==============================================")


# ------------------------------------------------------------
# 3. DISTRIBUCIÓN MUESTRAL DE LA MEDIA
# ------------------------------------------------------------

plt.figure(figsize=(12, 9))

for i, n in enumerate(tamanos_muestra):

    # --------------------------------------------------------
    # Generamos 10000 muestras de tamaño n
    # Cada dato procede de Binomial(N,p)
    # --------------------------------------------------------

    muestras = np.random.binomial(
        N,
        p,
        size=(numero_muestras, n)
    )

    # --------------------------------------------------------
    # Calculamos la media de CADA muestra
    # --------------------------------------------------------

    medias_muestrales = np.mean(muestras, axis=1)


    # --------------------------------------------------------
    # 4. PARÁMETROS EXPERIMENTALES
    # --------------------------------------------------------

    media_experimental = np.mean(medias_muestrales)

    varianza_experimental = np.var(
        medias_muestrales,
        ddof=1
    )


    # --------------------------------------------------------
    # 5. PARÁMETROS TEÓRICOS DE LA MEDIA MUESTRAL
    #
    # E(Xbarra) = mu
    #
    # Var(Xbarra) = sigma²/n
    # --------------------------------------------------------

    media_teorica = mu

    varianza_teorica = varianza / n

    desviacion_teorica = np.sqrt(varianza_teorica)


    # --------------------------------------------------------
    # MOSTRAMOS LOS RESULTADOS
    # --------------------------------------------------------

    print("\n==============================================")
    print("Tamaño de muestra n =", n)

    print("Media experimental =", media_experimental)
    print("Media teórica =", media_teorica)

    print("Varianza experimental =", varianza_experimental)
    print("Varianza teórica =", varianza_teorica)

    print("Diferencia de medias =",
          abs(media_experimental - media_teorica))

    print("Diferencia de varianzas =",
          abs(varianza_experimental - varianza_teorica))


    # --------------------------------------------------------
    # 6. REPRESENTACIÓN GRÁFICA
    # --------------------------------------------------------

    plt.subplot(2, 2, i + 1)

    # Histograma de las medias muestrales
    plt.hist(
        medias_muestrales,
        bins=30,
        density=True,
        alpha=0.7,
        label="Medias muestrales"
    )

    # --------------------------------------------------------
    # Normal teórica correspondiente a la distribución
    # muestral de la media
    # --------------------------------------------------------

    x = np.linspace(
        np.min(medias_muestrales),
        np.max(medias_muestrales),
        300
    )

    y = norm.pdf(
        x,
        loc=media_teorica,
        scale=desviacion_teorica
    )

    plt.plot(
        x,
        y,
        linewidth=2,
        label="Normal teórica"
    )

    plt.title(
        "Binomial N={}, p={}\n"
        "Tamaño muestra n={}".format(N, p, n)
    )

    plt.xlabel("Media muestral")
    plt.ylabel("Densidad")

    plt.grid()
    plt.legend()


plt.tight_layout()
plt.show()
```

```python
# ============================================================
# TURNO 9
# FUNCIÓN DE VEROSIMILITUD Y SOPORTE PARA UNA NORMAL
# Varianza conocida sigma^2 = 1
# Parámetro desconocido: mu
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# Semilla para reproducir los resultados
np.random.seed(1111111)

# ------------------------------------------------------------
# 1. PARÁMETROS REALES DE LA DISTRIBUCIÓN
# ------------------------------------------------------------

mu_real = 2
sigma = 1
n = 10

# Generamos una muestra de N(mu=2, sigma^2=1)
muestra = np.random.normal(
    loc=mu_real,
    scale=sigma,
    size=n
)

print("============================================")
print("MUESTRA GENERADA")
print("============================================")
print(muestra)

# Media de la muestra
media_muestral = np.mean(muestra)

print("\nMedia real mu =", mu_real)
print("Media de la muestra =", media_muestral)


# ------------------------------------------------------------
# 2. POSIBLES VALORES DEL PARÁMETRO mu
# ------------------------------------------------------------

MU = np.linspace(
    mu_real - 2,
    mu_real + 2,
    500
)


# ------------------------------------------------------------
# 3. FUNCIÓN DE VEROSIMILITUD
#
# L(mu) = producto de las densidades de cada observación
# ------------------------------------------------------------

verosimilitud = []

for mu in MU:

    # Densidad normal para cada dato de la muestra
    densidades = (
        1 / (np.sqrt(2 * np.pi) * sigma)
    ) * np.exp(
        -(muestra - mu)**2 / (2 * sigma**2)
    )

    # Producto de todas las densidades
    L = np.prod(densidades)

    verosimilitud.append(L)

verosimilitud = np.array(verosimilitud)


# ------------------------------------------------------------
# 4. FUNCIÓN SOPORTE = LOG-VEROSIMILITUD
# ------------------------------------------------------------

log_verosimilitud = []

for mu in MU:

    l_mu = (
        -n / 2 * np.log(2 * np.pi * sigma**2)
        - np.sum((muestra - mu)**2) / (2 * sigma**2)
    )

    log_verosimilitud.append(l_mu)

log_verosimilitud = np.array(log_verosimilitud)


# ------------------------------------------------------------
# 5. BUSCAMOS EL VALOR DE mu QUE MAXIMIZA LAS FUNCIONES
# ------------------------------------------------------------

indice_max_L = np.argmax(verosimilitud)
mu_max_L = MU[indice_max_L]

indice_max_logL = np.argmax(log_verosimilitud)
mu_max_logL = MU[indice_max_logL]


print("\n============================================")
print("ESTIMACIÓN POR MÁXIMA VEROSIMILITUD")
print("============================================")

print("Mu real =", mu_real)
print("Media muestral =", media_muestral)
print("Mu que maximiza L(mu) =", mu_max_L)
print("Mu que maximiza log L(mu) =", mu_max_logL)


# ------------------------------------------------------------
# 6. REPRESENTACIÓN GRÁFICA
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))


# ------------------------------------------------------------
# GRÁFICA 1: VEROSIMILITUD
# ------------------------------------------------------------

plt.subplot(1, 2, 1)

plt.plot(
    MU,
    verosimilitud,
    linewidth=2
)

# Valor verdadero de mu
plt.axvline(
    mu_real,
    linestyle="--",
    label="μ real = 2"
)

# Estimación máxima verosimilitud
plt.axvline(
    media_muestral,
    linestyle=":",
    label="μ estimado = media muestral"
)

plt.xlabel("μ")
plt.ylabel("L(μ)")
plt.title("Función de verosimilitud")
plt.grid()
plt.legend()


# ------------------------------------------------------------
# GRÁFICA 2: FUNCIÓN SOPORTE
# ------------------------------------------------------------

plt.subplot(1, 2, 2)

plt.plot(
    MU,
    log_verosimilitud,
    linewidth=2
)

plt.axvline(
    mu_real,
    linestyle="--",
    label="μ real = 2"
)

plt.axvline(
    media_muestral,
    linestyle=":",
    label="μ estimado = media muestral"
)

plt.xlabel("μ")
plt.ylabel("log L(μ)")
plt.title("Función soporte o log-verosimilitud")
plt.grid()
plt.legend()


plt.tight_layout()
plt.show()
```

```python
# ============================================================
# TURNO 9 - SEGUNDA PARTE
# ¿POR QUÉ EL MÁXIMO DE VEROSIMILITUD DE POISSON
# NO COINCIDE EXACTAMENTE CON lambda = 1?
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1111111)

# Parámetro verdadero
lambda_real = 1

# Diferentes tamaños de muestra
tamanos = [10, 50, 500, 5000]

# Posibles valores de lambda
lambdas = np.linspace(0.1, 2.5, 400)

plt.figure(figsize=(12, 9))

for i, n in enumerate(tamanos):

    # --------------------------------------------------------
    # 1. Generamos una muestra Poisson(lambda=1)
    # --------------------------------------------------------
    
    muestra = np.random.poisson(
        lambda_real,
        size=n
    )

    # Media muestral
    media_muestral = np.mean(muestra)

    # --------------------------------------------------------
    # 2. Calculamos la log-verosimilitud
    #
    # Ignoramos el término -log(x!) porque no depende
    # de lambda y, por tanto, no cambia dónde está el máximo.
    #
    # l(lambda) =
    # -n*lambda + sum(x_i)*log(lambda) + constante
    # --------------------------------------------------------

    logL = []

    for lam in lambdas:

        valor = (
            -n * lam
            + np.sum(muestra) * np.log(lam)
        )

        logL.append(valor)

    logL = np.array(logL)

    # --------------------------------------------------------
    # 3. Buscamos el lambda que maximiza la función
    # --------------------------------------------------------

    indice_maximo = np.argmax(logL)
    lambda_estimado = lambdas[indice_maximo]

    # --------------------------------------------------------
    # RESULTADOS
    # --------------------------------------------------------

    print("============================================")
    print("Tamaño de muestra n =", n)
    print("Lambda real =", lambda_real)
    print("Media muestral =", media_muestral)
    print("Lambda de máxima verosimilitud =", lambda_estimado)
    print("Diferencia respecto a lambda=1 =",
          abs(lambda_estimado - lambda_real))

    # --------------------------------------------------------
    # 4. GRÁFICA
    # --------------------------------------------------------

    plt.subplot(2, 2, i + 1)

    plt.plot(
        lambdas,
        logL,
        linewidth=2,
        label="Log-verosimilitud"
    )

    # Lambda verdadero
    plt.axvline(
        lambda_real,
        linestyle="--",
        label="λ real = 1"
    )

    # Lambda estimado
    plt.axvline(
        lambda_estimado,
        linestyle=":",
        label="λ estimado"
    )

    plt.xlabel("λ")
    plt.ylabel("log L(λ)")

    plt.title(
        "Poisson λ=1\n"
        "n={}, λ estimado={:.3f}".format(
            n,
            lambda_estimado
        )
    )

    plt.grid()
    plt.legend()


plt.tight_layout()
plt.show()


# ============================================================
# RESULTADO TEÓRICO
# ============================================================

print("\n============================================")
print("RESULTADO TEÓRICO")
print("============================================")

print("Para una distribución Poisson:")
print("Estimador MV de lambda = media muestral")
print("λ_hat = X_barra")
print("")
print("Cuando n -> infinito:")
print("X_barra -> lambda")
print("Por tanto, λ_hat -> 1")
```

### Conclusión del Turno 9

Para una distribución normal $N(\mu,1)$, se ha estudiado la función de verosimilitud y su función soporte o log-verosimilitud para estimar el parámetro desconocido $\mu$.

Se observa que ambas funciones alcanzan su máximo para el mismo valor de $\mu$. Esto se debe a que el logaritmo es una función monótona creciente y, por tanto, transformar la verosimilitud mediante el logaritmo no modifica la posición de su máximo.

El estimador de máxima verosimilitud de la media de una distribución normal es:

$$
\hat{\mu}=\bar X
$$

Por ello, aunque la muestra se haya generado utilizando un valor real $\mu=2$, el máximo de la función de verosimilitud no tiene por qué encontrarse exactamente en 2, sino en la media obtenida en la muestra concreta.

El mismo razonamiento explica por qué, en el ejemplo de la distribución de Poisson con parámetro real $\lambda=1$, el valor más verosímil de $\lambda$ no coincide necesariamente de forma exacta con 1.

Para una distribución de Poisson, el estimador de máxima verosimilitud es:

$$
\hat{\lambda}=\bar X
$$

Como la media de una muestra aleatoria finita no tiene por qué coincidir exactamente con la media poblacional, puede obtenerse $\hat{\lambda}\neq1$.

En la simulación realizada se observa que, para muestras pequeñas, el valor estimado puede alejarse de 1. Sin embargo, al aumentar el tamaño de la muestra, la estimación se aproxima progresivamente al parámetro verdadero. Por ejemplo, para $n=5000$ se obtiene aproximadamente $\hat{\lambda}=0.990$.

Por la Ley de los Grandes Números:

$$
\bar X\rightarrow E(X)=\lambda
$$

cuando:

$$
n\rightarrow\infty
$$

y, por tanto:

$$
\boxed{\hat{\lambda}\rightarrow1}
$$

En conclusión, la diferencia entre el parámetro verdadero y el valor de máxima verosimilitud en una muestra finita se debe a la variabilidad muestral y no a un error del método.

### Conclusión del Turno 10

Se ha estudiado cómo varía la precisión de la función de verosimilitud y de la función soporte de una distribución normal en función del tamaño de la muestra y de los parámetros de la distribución.

Para una variable aleatoria normal con media desconocida $\mu$ y varianza conocida $\sigma^2$, el estimador de máxima verosimilitud de la media es:

$$
\hat{\mu}=\bar X
$$

y su varianza es:

$$
Var(\hat{\mu})=\frac{\sigma^2}{n}
$$

#### Influencia del tamaño de la muestra

Manteniendo fija la varianza $\sigma^2=1$, se han estudiado diferentes tamaños muestrales.

Al aumentar $n$, la función de verosimilitud se hace progresivamente más estrecha y concentrada alrededor de su máximo. Del mismo modo, la función soporte presenta una curvatura cada vez mayor alrededor del máximo.

Esto indica un aumento de la precisión en la estimación de $\mu$, ya que:

$$
Var(\hat{\mu})=\frac{1}{n}
$$

y, por tanto:

$$
n\rightarrow\infty
\quad\Rightarrow\quad
Var(\hat{\mu})\rightarrow0
$$

Así, cuanto mayor es el tamaño de la muestra, mayor es la información disponible sobre el parámetro y más precisa es su estimación.

#### Influencia de la varianza

Manteniendo fijo el tamaño de muestra $n=50$, se han estudiado distintos valores de $\sigma$.

La varianza del estimador es:

$$
Var(\hat{\mu})=\frac{\sigma^2}{50}
$$

Los valores teóricos obtenidos son:

- Para $\sigma=0.5$: $Var(\hat{\mu})=0.005$
- Para $\sigma=1$: $Var(\hat{\mu})=0.020$
- Para $\sigma=2$: $Var(\hat{\mu})=0.080$
- Para $\sigma=4$: $Var(\hat{\mu})=0.320$

Gráficamente se observa que, cuando aumenta $\sigma$, la función de verosimilitud se hace más ancha y menos concentrada alrededor del máximo. Por tanto, aumenta la incertidumbre en la estimación de $\mu$.

En conclusión:

$$
\boxed{n\uparrow\Rightarrow Var(\hat{\mu})\downarrow
\Rightarrow\text{mayor precisión}}
$$

mientras que:

$$
\boxed{\sigma^2\uparrow\Rightarrow Var(\hat{\mu})\uparrow
\Rightarrow\text{menor precisión}}
$$

La función de verosimilitud y la función soporte alcanzan su máximo en el mismo valor del parámetro. La función soporte facilita el análisis matemático y gráfico, mientras que ambas contienen la misma información respecto al valor de máxima verosimilitud.

Por tanto, la precisión en la estimación de la media de una distribución normal mejora al aumentar el tamaño de la muestra y empeora al aumentar la variabilidad de la población.
